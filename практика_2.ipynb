```python
from collections import defaultdict
d = defaultdict(list)
d['a'].append(1)
d['a'].append(2)
d['b'].append(3)
print(d)
```

```python
from collections import Counter
words = ['py','js','py','py','js']
cnt = Counter(words)
print(cnt)
# Counter({'py':3,'js':2})
print(cnt.most_common(1))
# [('py', 3)]
```

```python
from collections import namedtuple
Point = namedtuple('Point', ['x','y'])
p = Point(10, 20)
print(p.x)
print(p.y)
print(p)
# Point(x=10, y=20)
```

```python
from collections import deque
q = deque([1, 2, 3])

# Point(x=10, y=20)
q.appendleft(0) # зліва
q.append(4) # справа
q.popleft() # прибрати зліва
print(q)
```

```python
from itertools import groupby

data = [('fruit', 'apple'), ('veg', 'carrot'), ('fruit', 'pear')]

data.sort(key=lambda x: x[0])

for k, g in groupby(data, lambda x: x[0]):
    print(k, list(g))
```

```python
from itertools import combinations
products = ['Laptop','Phone','Tablet','Watch']
bundles = list(combinations(products, 2))
print(len(bundles), 'пар')
print(bundles[:3])
```

```python
from itertools import chain
q1 = [100, 200, 150]
q2 = [180, 220, 190]
all_sales = list(chain(q1, q2))
print(all_sales)
print(sum(all_sales))
```
